# KNN:

- k nearest neighbors algorithem:
- The knn algorithem is a supervised machine learning algorithem perform on classification and regression problem.It also handle both numerical and categorical data so it perform on both classification and regression data sets.
- It use for recognition ,data mining and intrusion detection.
- the KNN works by finding the neirest nieghbers to given data point based on distance..

### Advantages:

- Easy to implement.
- Adapts Easily.
- few Hyperparamter

### Disadvatages :
- Dose not scale
- Cures of dimentionality 
- Prone to overfiting

```python
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
```

```python
x=df.drop(['target','Flower_name'],axis=1)
y=df['target']
x
```

```python
y
```

```python
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.4,random_state=23)
knn = KNeighborsClassifier()
knn.fit(x_train,y_train)
```

```python
y_pred=knn.predict(x_test)
```

```python
knn.score(x_test,y_test)
```

```python
knn.predict([[4.6,3.1,1.5,0.2]])
```

```python
from sklearn.metrics import confusion_matrix

cm=confusion_matrix(y_pred,y_test)
cm
```

```python
plt.figure(figsize=(6,3))
sns.heatmap(cm,annot=True)
plt.xlabel('Predicted')
plt.ylabel('Actual')
```

```python
from sklearn.metrics import classification_report

print(classification_report(y_pred,y_test))
```

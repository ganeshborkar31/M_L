```python
class animal:
    def __init__(self,name,color):
        self.name = name
        self.color = color
        
    def run(self):
        print( f'{self.name} can Run')
    
class dog (animal):
    
    def bark(self):
        return 'Dog is barking'
    
d1 = dog('dog','white')
```

```python
d1.run()
```

```python
d1.bark()
```

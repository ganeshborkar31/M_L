### 4 Explain and perform knn?

#### KNN - K nearest neighbors - 
- KNN is the supervised ML algorithem used for the classification problem.
- using KNN we can classify the New data based on old data that is used for the taining


```python
import pandas as pd
import numpy as np
import seaborn as sns
from sklearn.neighbors import KNeighborsClassifier
```

```python
from sklearn.datasets import load_iris
iris = load_iris()
df = pd.DataFrame(iris.data,columns=iris.feature_names)
df['target'] = iris.target
df
```

```python
X=df.drop('target',axis=1)
y=df['target']
from sklearn.model_selection import train_test_split

xtrain,xtest,ytrain,ytest = train_test_split(X,y,test_size=0.3,random_state=32)
ytrain
```

```python
knn=KNeighborsClassifier(3)
knn.fit(xtrain,ytrain)
```

```python
knn.score(xtest,ytest)
```

## Entropy:
- $E(s) = \sum_{i=1}^{e}{P}_i \log{P_i}.$
- ${P_i}$ is the frequentist probablity of element/class "i" in our data 

- More the uncertainty more the entropy.
- for the 2 class problem minimum Entropy is 0 and maximum is 1.
- For the more than 2 classes the min Entropy is 0 but max Entropy can be greter than one.
- Both $log_2$ and $log_e$ are used to calculate Entorpy.
